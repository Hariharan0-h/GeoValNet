```python
import pandas as pd

df = pd.read_csv("../data/sample_houses.csv")

median_price = (
    df.groupby("zipcode")["price"]
      .median()
      .reset_index()
)

median_price.head()
```

```python
import folium
from pathlib import Path
```

```python
# Seattle city center (placeholder)
center = [47.6062, -122.3321]

# Create interactive map
m = folium.Map(
    location=center,
    zoom_start=10,
    tiles="OpenStreetMap"
)

# Sample high-value ZIP code locations
locations = [
    {"zip": "98039", "price": "$2.1M", "lat": 47.6205, "lon": -122.3493},
    {"zip": "98004", "price": "$1.8M", "lat": 47.6101, "lon": -122.2015},
    {"zip": "98040", "price": "$1.7M", "lat": 47.5707, "lon": -122.2221},
]

for loc in locations:
    folium.CircleMarker(
        location=[loc["lat"], loc["lon"]],
        radius=8,
        popup=f"ZIP: {loc['zip']}<br>Median Price: {loc['price']}",
        color="red",
        fill=True,
        fill_opacity=0.7,
    ).add_to(m)

m
```

```python
# Export map
Path("../reports").mkdir(exist_ok=True)

m.save("../reports/map.html")

print("Map exported successfully to reports/map.html")
```
