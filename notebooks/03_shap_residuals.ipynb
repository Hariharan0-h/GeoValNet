```python
import pandas as pd
import numpy as np
import shap

from xgboost import XGBRegressor
```

```python
X = pd.DataFrame({
    "dist_city_center": [2.1,4.3,6.2,8.1,1.5],
    "sqft_living":[1500,2200,1800,2600,3200],
    "bedrooms":[3,4,3,5,4]
})

y = [420000,610000,530000,780000,920000]
```

```python
model = XGBRegressor(
    random_state=42,
    n_estimators=50,
    max_depth=3
)

model.fit(X,y)
```

```python
explainer = shap.Explainer(model)

shap_values = explainer(X)
```

```python
shap.plots.beeswarm(shap_values)
```
