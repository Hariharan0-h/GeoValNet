```python
import pandas as pd
import numpy as np
import shap

from xgboost import XGBRegressor
```

```python
X = pd.DataFrame({
    "dist_city_center": [2.1,4.3,6.2,8.1,1.5],
    "sqft_living":[1500,2200,1800,2600,3200],
    "bedrooms":[3,4,3,5,4]
})

y = [420000,610000,530000,780000,920000]
```

```python
model = XGBRegressor(
    random_state=42,
    n_estimators=50,
    max_depth=3
)

model.fit(X,y)
```

```python
explainer = shap.Explainer(model)

shap_values = explainer(X)
```

```python
shap.plots.beeswarm(shap_values)
```

```python
import os
import matplotlib.pyplot as plt

os.makedirs("reports/figures", exist_ok=True)

# Save beeswarm plot
shap.plots.beeswarm(shap_values, show=False)
plt.savefig("reports/figures/shap_beeswarm.png",
            dpi=300,
            bbox_inches="tight")
plt.close()

# Save bar plot
shap.plots.bar(shap_values, show=False)
plt.savefig("reports/figures/shap_summary.png",
            dpi=300,
            bbox_inches="tight")
plt.close()

print("SHAP figures exported successfully.")
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Predictions
predictions = model.predict(X)

# Residuals
residuals = y - predictions

# Create DataFrame
results = pd.DataFrame({
    "Actual": y,
    "Predicted": predictions,
    "Residual": residuals
})

print(results)
```

```python
plt.figure(figsize=(7,5))

plt.scatter(predictions, residuals)

plt.axhline(0, color="red", linestyle="--")

plt.xlabel("Predicted Price")
plt.ylabel("Residual")

plt.title("Residual Plot")

plt.show()
```

## Residual Analysis

The residual plot indicates that the baseline XGBoost model tends to
underestimate high-value luxury and waterfront properties.

This suggests that additional spatial information can improve predictions
for premium housing markets.
