```python
SEATTLE_LAT = 47.6062
SEATTLE_LONG = -122.3321
```

```python
def haversine_distance(lat1, lon1, lat2, lon2):

    R = 6371  # Earth radius in kilometers

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2)**2 +
        np.cos(lat1) *
        np.cos(lat2) *
        np.sin(dlon / 2)**2
    )

    c = 2 * np.arcsin(np.sqrt(a))

    return R * c
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/king_county.csv")

df.head()
```

```python
SEATTLE_LAT = 47.6062
SEATTLE_LONG = -122.3321
```

```python
def haversine_distance(lat1, lon1, lat2, lon2):

    R = 6371

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat/2)**2 +
        np.cos(lat1) *
        np.cos(lat2) *
        np.sin(dlon/2)**2
    )

    c = 2 * np.arcsin(np.sqrt(a))

    return R * c
```

```python
df["dist_city_center_km"] = haversine_distance(
    df["lat"],
    df["long"],
    SEATTLE_LAT,
    SEATTLE_LONG
)
```

```python
df.head()
```

```python
df.to_csv("../data/king_county_with_geo_features.csv", index=False)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df["long"],
    df["lat"]
)

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Property Distribution in King County")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df["long"],
    df["lat"],
    c=df["price"],
    alpha=0.7
)

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Property Prices by Location")

plt.colorbar(label="Price")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df["dist_city_center_km"],
    df["price"]
)

plt.xlabel("Distance from Seattle City Center (km)")
plt.ylabel("Price")
plt.title("Property Price vs Distance from City Center")

plt.show()
```
