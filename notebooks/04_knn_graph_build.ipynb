```python
import pandas as pd
import numpy as np
from sklearn.neighbors import BallTree
```

```python
# Bellevue sample coordinates
df = pd.DataFrame({
    "lat": [
        47.6101,
        47.6112,
        47.6123,
        47.6135,
        47.6140,
        47.6152,
        47.6161,
        47.6174,
        47.6182,
        47.6191,
        47.6200
    ],
    "long": [
        -122.2015,
        -122.2021,
        -122.2030,
        -122.2040,
        -122.2051,
        -122.2063,
        -122.2074,
        -122.2080,
        -122.2091,
        -122.2102,
        -122.2110
    ]
})

df.head()
```

```python
coords = np.radians(df[["lat", "long"]])

tree = BallTree(coords, metric="haversine")

distances, indices = tree.query(coords, k=10)
```

```python
# Convert radians to kilometres
distances_km = distances * 6371

print("Neighbour indices")
print(indices)

print()

print("Distances (km)")
print(distances_km)
```

```python
print("Maximum neighbour distance:",
      round(distances_km.max(), 2),
      "km")
```

## Validation

The Bellevue sample confirms that the BallTree KNN graph correctly
identifies the nearest neighbours.

For K=10, all neighbouring properties lie within approximately
2 kilometres, validating the spatial graph construction.

## Validation

The Bellevue sample confirms that the BallTree KNN graph correctly
identifies neighbouring properties.

For K=10, the neighbours are expected to lie within approximately
2 kilometres, validating the graph construction.

```python
import folium

# Center map on Bellevue sample
center = [df["lat"].mean(), df["long"].mean()]

m = folium.Map(
    location=center,
    zoom_start=14
)
```

```python
for _, row in df.iterrows():
    folium.CircleMarker(
        location=[row["lat"], row["long"]],
        radius=5,
        color="blue",
        fill=True,
        fill_opacity=0.7,
    ).add_to(m)

m
```

```python
for i in range(len(df)):
    for neighbour in indices[i][1:]:   # Skip self

        folium.PolyLine(
            locations=[
                [df.iloc[i]["lat"], df.iloc[i]["long"]],
                [df.iloc[neighbour]["lat"], df.iloc[neighbour]["long"]]
            ],
            color="red",
            weight=1.5,
            opacity=0.6
        ).add_to(m)

m
```

```python
m.save("reports/figures/knn_graph_map.html")

print("Map saved successfully.")
```

```python
import pandas as pd
import numpy as np

from sklearn.neighbors import NearestNeighbors

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/raw/kc_house_data.csv")

df.head()
```

```python
coords = df[["lat", "long"]]

coords.head()
```

```python
K = 10

knn = NearestNeighbors(
    n_neighbors=K,
    metric="euclidean"
)

knn.fit(coords)
```

```python
distances, indices = knn.kneighbors(coords)
```

```python
print(distances.shape)
print(indices.shape)
```

```python
print("Neighbour indices")

print(indices[0])

print()

print("Distances")

print(distances[0])
```
