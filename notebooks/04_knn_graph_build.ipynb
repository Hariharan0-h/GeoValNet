```python
import pandas as pd
import numpy as np
from sklearn.neighbors import BallTree
```

```python
# Bellevue sample coordinates
df = pd.DataFrame({
    "lat": [
        47.6101,
        47.6112,
        47.6123,
        47.6135,
        47.6140,
        47.6152,
        47.6161,
        47.6174,
        47.6182,
        47.6191,
        47.6200
    ],
    "long": [
        -122.2015,
        -122.2021,
        -122.2030,
        -122.2040,
        -122.2051,
        -122.2063,
        -122.2074,
        -122.2080,
        -122.2091,
        -122.2102,
        -122.2110
    ]
})

df.head()
```

```python
coords = np.radians(df[["lat", "long"]])

tree = BallTree(coords, metric="haversine")

distances, indices = tree.query(coords, k=10)
```

```python
# Convert radians to kilometres
distances_km = distances * 6371

print("Neighbour indices")
print(indices)

print()

print("Distances (km)")
print(distances_km)
```

```python
print("Maximum neighbour distance:",
      round(distances_km.max(), 2),
      "km")
```

## Validation

The Bellevue sample confirms that the BallTree KNN graph correctly
identifies the nearest neighbours.

For K=10, all neighbouring properties lie within approximately
2 kilometres, validating the spatial graph construction.

```python
coords = np.radians(df[["lat", "long"]])

tree = BallTree(coords, metric="haversine")

distances, indices = tree.query(coords, k=10)
```

```python
# Convert radians to kilometres
distances_km = distances * 6371

print("Neighbour Indices")
print(indices)

print("\nDistances (km)")
print(distances_km)
```

```python
print(f"Maximum neighbour distance: {distances_km.max():.2f} km")

if distances_km.max() < 2:
    print("✅ Validation Passed: All K=10 neighbours are within 2 km.")
else:
    print("⚠️ Some neighbours are farther than 2 km.")
```

identifies neighbouring properties.

For K=10, the neighbours are expected to lie within approximately
