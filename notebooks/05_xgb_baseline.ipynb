# Week 2 Update

After integrating the Point of Interest (POI) distance features contributed by
the R2 team, the XGBoost baseline model is re-evaluated using 5-fold cross
validation.

New spatial features include:

- Distance to nearest school
- Distance to nearest hospital

The updated feature set is expected to improve predictive performance by
providing additional geographic context.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_percentage_error
from sklearn.metrics import root_mean_squared_error
from xgboost import XGBRegressor

df = pd.DataFrame({
    "sqft":[1200,1800,2200,2600,3100,3400,3900,4100],
    "school_dist":[2.1,1.5,0.8,3.2,2.8,1.0,0.5,2.3],
    "hospital_dist":[4.2,3.6,2.8,5.1,4.3,2.5,1.8,3.4],
    "price":[320000,470000,560000,670000,760000,860000,970000,1080000]
})

X=df.drop(columns="price")
y=df["price"]

kf=KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

model=XGBRegressor(
    random_state=42,
    n_estimators=100
)

mape=[]
rmse=[]

for train,test in kf.split(X):

    model.fit(X.iloc[train],y.iloc[train])

    pred=model.predict(X.iloc[test])

    mape.append(
        mean_absolute_percentage_error(
            y.iloc[test],
            pred
        )
    )

    rmse.append(
        root_mean_squared_error(
            y.iloc[test],
            pred
        )
    )

print("Average MAPE",np.mean(mape))
print("Average RMSE",np.mean(rmse))
```
