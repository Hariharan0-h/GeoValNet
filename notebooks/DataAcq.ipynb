## Check House Price Distribution

```python
# Visualization
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["price"], bins=40)

plt.title("House Price Distribution")

plt.show()
```

The above visualisation says that the distibution is right skewed.

## Detect Outliers

```python
plt.figure(figsize=(10,4))

sns.boxplot(x=df["price"])

plt.title("Price Outliers")

plt.show()
```

## Remove Extreme Outliers (IQR Method)

```python
Q1 = df["price"].quantile(0.25)

Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR

upper = Q3 + 1.5*IQR

df_clean = df[(df["price"]>=lower) &
              (df["price"]<=upper)]

print(df_clean.shape)
```

```python
df_clean.head()
```

## Normalize Prices

```python
import numpy as np
```

```python
df_clean["price_log"] = np.log1p(df_clean["price"])
```

## Compare before and after normalisation

```python
fig, ax = plt.subplots(1,2,figsize=(12,4))

sns.histplot(df_clean["price"], ax=ax[0])

ax[0].set_title("Original Price")

sns.histplot(df_clean["price_log"], ax=ax[1])

ax[1].set_title("Log Normalized Price")

plt.show()
```

## Convert to GeoDataFrame

```python
# Geospatial
import geopandas as gpd
from shapely.geometry import Point
```

```python
geometry = gpd.points_from_xy(df_clean["long"],
                              df_clean["lat"])

gdf = gpd.GeoDataFrame(df_clean,
                       geometry=geometry)

gdf.head()
```

```python
gdf['geometry'].head()
```

## Set Coordinate Reference System (CRS)

```python
gdf.crs = "EPSG:4326"

print(gdf.crs)
```

## Plot Houses using GeoPandas

```python
gdf.plot(figsize=(10,10), markersize=1)

plt.title("Geographical Distribution of Houses")

plt.show()
```

## Interactive Map using Folium

```python
# Interactive Maps
import folium
```

```python
centre_lat = gdf['lat'].mean()
centre_long = gdf['long'].mean()

house_map = folium.Map(location=[centre_lat, centre_long], Zoom_start=10)
```

```python
for _, row in gdf.head(100).iterrows():
    folium.CircleMarker(location=[row['lat'], row['long']],
                        radius=2,
                        color='blue',
                        fill=True,
                        fill_color='blue',
                        fill_opacity=0.7).add_to(house_map)
```

```python
house_map
```

## Distance Calculation

```python
from haversine import haversine, Unit
```

```python
house1 = (gdf.iloc[0]['lat'], gdf.iloc[0]['long'])
house2 = (gdf.iloc[1]['lat'], gdf.iloc[1]['long'])

distance = haversine(house1, house2, unit=Unit.KILOMETERS)

print(distance)
```
